```python
import pandas as pd
import os

os.makedirs('data/full', exist_ok=True)
os.makedirs('test/clean', exist_ok=True)

x = pd.read_csv('data/raw/hotele_raw.csv')
y = pd.read_csv('data/raw/reviews_raw.csv') 
z = pd.read_csv('data/raw/details_raw.csv')

print(f"""
hotele: ({x.shape}) {x.columns} \n\n
reviews: ({y.shape}) {y.columns} \n\n
details: ({z.shape}) {z.columns} \n\n
""")
```

```python
# =========================================
# HOTELE
# =========================================
x_keep = [
    "location_id",
    "name",
    # "query_city",
    "address_obj.city",
    "address_obj.state",
    # "address_obj.country",
]
x = x[[col for col in x_keep if col in x.columns]]
```

```python
# =========================================
# REVIEWS
# =========================================

y_keep = [
    "id",
    "location_id",
    "published_date",
    "rating",
    "helpful_votes",
    "text",
    "title",
    "trip_type",
    "travel_date",
    "user.username",
    "user.user_location.name",
    
    'owner_response.title', 'owner_response.text',
    'owner_response.author', 'owner_response.published_date',
    
    "subratings.0.localized_name","subratings.0.value",
    "subratings.1.localized_name","subratings.1.value",
    "subratings.2.localized_name","subratings.2.value",
    "subratings.3.localized_name","subratings.3.value",
    "subratings.4.localized_name","subratings.4.value",
    "subratings.5.localized_name","subratings.5.value",
]

y = y[[col for col in y_keep if col in y.columns]]
```

```python
# =========================================
# DETAILS
# =========================================

z_keep = [
    "location_id",
    "name",
    "latitude",
    "longitude",
    "rating",
    "num_reviews",
    "photo_count",
    "price_level",
    "amenities",
    "styles",
    "trip_types",
    "address_obj.city",
    "address_obj.country",
    "ranking_data.ranking_out_of",
    "ranking_data.ranking",
    "review_rating_count.1",
    "review_rating_count.2",
    "review_rating_count.3",
    "review_rating_count.4",
    "review_rating_count.5",
    # "category.localized_name",
    "brand",
    "description",
    "subratings.0.localized_name","subratings.0.value",
    "subratings.1.localized_name","subratings.1.value",
    "subratings.2.localized_name","subratings.2.value",
    "subratings.3.localized_name","subratings.3.value",
    "subratings.4.localized_name","subratings.4.value",
    "subratings.5.localized_name","subratings.5.value",
]

z = z[[col for col in z_keep if col in z.columns]]
```

```python
x.to_csv("data/hotele_short.csv", index=False)
x.head(2)

y.to_csv("data/reviews_short.csv", index=False)
y.head(2)

z.to_csv("data/details_short.csv", index=False)
z.head(2)
```

```python
test_x = x.head()
test_y = y.head()
test_z = z.head()


test_x.to_csv('test/clean/hotele_test.csv', index=False)
test_y.to_csv('test/clean/reviews_test.csv', index=False)
test_z.to_csv('test/clean/details_test.csv', index=False)
```
